```python
def read_gmt(file_path):
    gene_sets = {}
    with open(file_path, 'r') as f:
        for line in f:
            entries = line.strip().split('\t')
            gene_set_name = entries[0]
            description = entries[1]
            genes = entries[2:]
            gene_sets[gene_set_name] = {
                'description': description,
                'genes': genes
            }
    return gene_sets
```

```python
# up genes
deg_ft = deg[
    (deg['group'].isin(['BC'])) &
    (deg['FDR'] < 0.05)]


goi_init = deg_ft['gene'].unique()
len(goi_init)
```

Filter: for each of these genes in BC, NHW vs NHB, by mannwhitenyu, record P, do FDR correction

```python
from scipy.stats import mannwhitneyu
from statsmodels.stats.multitest import multipletests
import pandas as pd
import numpy as np

# Filter for BC group
bc_mask = meta['group'] == 'BC'
bc_meta = meta.loc[bc_mask]
bc_expr = expr.loc[goi_init, bc_mask]

pvals = []
mean1 = []
mean2 = []
ratio = []

for gene in goi_init:
    x = bc_expr.loc[gene, bc_meta['race'] == 'NHW']
    y = bc_expr.loc[gene, bc_meta['race'] == 'NHB']
    # If either group has no variance or not enough samples, skip
    if (len(x) < 2) or (len(y) < 2):
        pvals.append(np.nan)
        mean1.append(np.nan)
        mean2.append(np.nan)
        ratio.append(np.nan)
        continue
    stat, p = mannwhitneyu(x, y, alternative='two-sided')
    pvals.append(p)
    m1 = np.mean(x)
    m2 = np.mean(y)
    mean1.append(m1)
    mean2.append(m2)
    # Avoid division by zero
    ratio.append(m2/m1 if m1 != 0 else np.nan)

# FDR correction
pvals = np.array(pvals)
fdr_pvals = multipletests(pvals, method='fdr_bh')[1]

# Results DataFrame
results_df = pd.DataFrame({
    'gene': goi_init,
    'pval': pvals,
    'fdr': fdr_pvals,
    'mean_NHW': mean1,
    'mean_NHB': mean2,
    'NHB/NHW': ratio
}).sort_values('fdr')

results_dfsel = results_df[
    (results_df['fdr'] < 0.05) &
    (results_df['NHB/NHW'] > 1)
    ]
print(results_dfsel.shape)
results_dfsel
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

# ---- mapping ----
celltype_to_major = {}
for major, members in cells_group2.items():
    for ct in members:
        celltype_to_major[ct] = major

deg_sel = deg_sel.copy()
deg_sel['major'] = deg_sel['celltype'].map(celltype_to_major).fillna('Other')
deg_sel = deg_sel[deg_sel['major'] != 'Other']  # drop Other if needed

# ---- count per gene x major ----
counts = (
    deg_sel.groupby(['gene', 'major'], observed=True)
           .size()
           .rename('count')
           .reset_index()
)

# total per gene for sorting columns
gene_totals = counts.groupby('gene', observed=True)['count'].sum().sort_values(ascending=False)
genes_order = gene_totals.index.tolist()

# fixed major order
majors_all = ['CD4T', 'CD8T', 'other T', 'B', 'DC', 'Mono', 'NK']
majors_present = [m for m in majors_all if m in counts['major'].unique()]

# ---- plotting ----
fig, ax = plt.subplots(figsize=(3.2, 1.5))

x_positions = range(len(genes_order))

for xi, gene in zip(x_positions, genes_order):
    row = counts[counts['gene'] == gene][['major', 'count']].copy()
    missing = set(majors_present) - set(row['major'])
    if missing:
        row = pd.concat([row, pd.DataFrame({'major': list(missing), 'count': 0})], ignore_index=True)
    row['major_cat_order'] = row['major'].apply(lambda m: majors_present.index(m))
    row = row.sort_values(['count', 'major_cat_order'], ascending=[False, True])

    bottom = 0
    for _, r in row.iterrows():
        m, c = r['major'], int(r['count'])
        if c > 0:
            ax.bar(xi, c, bottom=bottom, color=major_celltype_palette.get(m, '#BBBBBB'),
                   edgecolor='none', width=0.7)
            bottom += c

# axis labels & ticks
ax.set_xticks(list(x_positions))
ax.set_xticklabels(genes_order, fontsize=5, rotation=90)
ax.set_ylabel('Count', fontsize=5)
ax.set_xlabel('', fontsize=5)
ax.tick_params(axis='x', labelsize=5, pad=-3)
ax.tick_params(axis='y', labelsize=5, length=1, pad=0)
ax.spines[['top', 'right']].set_visible(False)

# legend in upper right
handles = [Patch(color=major_celltype_palette[m], label=m) for m in majors_present]
leg = ax.legend(handles=handles, title='Major cell type', frameon=False,
                ncol=3, fontsize=5, title_fontsize=5, loc='upper right')

plt.tight_layout()
plt.savefig('plots/Figure8/UpGenes.svg')
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
from scipy.stats import zscore

# --- inputs ---
# dfp: DataFrame with expression + metadata rows (must contain rows 'race' and 'group')
# gois: list of gene symbols (rows) to plot as heatmap

# --- z-score genes across patients (columns) ---
expr_df = dfp.loc[gois].apply(pd.to_numeric, errors='coerce')  # ensure numeric
expr_z = pd.DataFrame(
    zscore(expr_df.values, axis=1, nan_policy='omit'),
    index=expr_df.index,
    columns=expr_df.columns
)

# --- build race-stage annotation (aligned to columns) ---
race  = dfp.loc['race',  expr_z.columns].astype(str)
group = dfp.loc['group', expr_z.columns].astype(str)
race_stage = (race + '-' + group)   # e.g., 'NHW-DCIS'

race_stage_order = ['NHW-HRL', 'NHW-DCIS', 'NHW-BC', 'NHB-HRL', 'NHB-DCIS', 'NHB-BC']
race_stage_cat = pd.Categorical(race_stage, categories=race_stage_order, ordered=True)
order_cols = race_stage_cat.argsort()

# Reorder the expression matrix and the col_colors
expr_z = expr_z.iloc[:, order_cols]
# col_colors = col_colors.iloc[order_cols]

palette = {
    'NHW-HRL':  '#B9DBF4',
    'NHW-DCIS': '#95AAD3',
    'NHW-BC':   '#3A68AE',
    'NHB-HRL':  '#EDAEAE',
    'NHB-DCIS': '#E07F80',
    'NHB-BC':   '#C25757',
}
col_colors = race_stage.map(palette)

# --- clustermap ---
g = sns.clustermap(
    expr_z,
    cmap='vlag', center=0,
    col_colors=col_colors,
    row_cluster=True, col_cluster=False,
    cbar_kws={'shrink': 0.5},
    figsize=(4.5,3.5),
    dendrogram_ratio=(0.06, 0.06),
    colors_ratio=0.02
)

# --- tidy ticks: show all gene names (clustered order), no x labels ---
# use the *reordered* index from the clustered matrix
ordered_genes = g.data2d.index.tolist()
n_rows = len(ordered_genes)

# 1) put one tick per row (clustermap/heatmap rows are centered at 0.5, 1.5, ..., n-0.5)
g.ax_heatmap.set_yticks(np.arange(n_rows) + 0.5)
# 2) set labels to the reordered gene names
g.ax_heatmap.set_yticklabels(ordered_genes, fontsize=5, rotation=0)

# remove all x tick labels (patients)
g.ax_heatmap.set_xticks([])          # no ticks
g.ax_heatmap.set_xticklabels([])     # no labels

# cosmetic tweaks (no “stick-out” ticks)
g.ax_heatmap.tick_params(axis='y', length=0, pad=1, labelsize=5)
g.ax_heatmap.tick_params(axis='x', length=0, pad=0, labelsize=5)

g.ax_col_colors.set_xticks([])
g.ax_col_colors.set_yticks([])
g.ax_col_colors.tick_params(axis='both', length=0)

# --- legend for race-stage annotation ---
handles = [Patch(facecolor=palette[k], label=k) for k in palette]
g.ax_heatmap.legend(
    handles=handles,
    title='',
    frameon=False,
    fontsize=5, title_fontsize=5,
    loc='upper left',
    bbox_to_anchor=(1.12, 1.0)  # place just outside heatmap
)

cax = g.cax
cax.set_position([.98, .15
                  , 0.01, 0.1
                 ])  
cax.set_title("z-score", fontsize=5, pad=2)
g.fig.savefig('plots/Figure8/NHBUpSig_heatmap_exprz_with_racestage.svg', bbox_inches='tight')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import textwrap

# --- figure ---
fig, ax = plt.subplots(figsize=(3,2.7))

dfp2['Term'] = [x.split(' (GO')[0] for x in dfp2['Term']]

# Barplot (no error bars)
sns.barplot(
    data=dfp2.iloc[:20,:],
    x="Combined Score",     # numeric values
    y="Term",          # categories
    orient='h',        # vertical bars
    color='salmon',
    errorbar=None,
    ax=ax
)

# Format xtick labels: remove underscores + wrap into 2–3 words per line
def wrap_label(label, width=50):
    """Wrap text to given width for tick labels."""
    return "\n".join(textwrap.wrap(label.replace("_", " "), width=width))

ax.set_yticklabels(
    [wrap_label(t.get_text()) for t in ax.get_yticklabels()],
    rotation=0, fontsize=5#, ha='right', rotation_mode='anchor'
)
ax.tick_params(axis='y', labelsize=5, pad=-1)
ax.tick_params(axis='x', labelsize=5, pad=-1)

# Remove legend if present
if ax.get_legend() is not None:
    ax.legend_.remove()

# Axis labels & title
ax.set_xlabel('Combined score', fontsize=6)
ax.set_ylabel('')
ax.set_title('', fontsize=7, weight='bold')

plt.tight_layout()
plt.savefig('plots/Figure8/NHBUp_enr.svg', bbox_inches='tight')
plt.show()
```
