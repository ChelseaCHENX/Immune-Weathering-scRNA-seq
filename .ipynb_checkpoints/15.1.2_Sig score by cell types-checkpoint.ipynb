````python
def read_gmt(file_path):
    gene_sets = {}
    with open(file_path, 'r') as f:
        for line in f:
            entries = line.strip().split('\t')
            gene_set_name = entries[0]
            description = entries[1]
            genes = entries[2:]
            gene_sets[gene_set_name] = {
                'description': description,
                'genes': genes
            }
    return gene_sets
````

# Load data

````python
scores = pd.read_csv(
    'data/08_signature/ssgsea_NHBSig_IWSCdataset.csv',
    index_col=0
)

scores
````

````python
scores['dataset_name'] = [
    x.replace('-','_') for x in scores['dataset_name']
]
````

````python
meta = pd.read_csv(
    'data/01_meta/meta_pt.csv',
    index_col=0
)
meta
````

````python
mg = pd.merge(
    scores, meta, on='dataset_name'
)

mg['race']
````

````python
mg['race_group'] = mg['race'] + '-' + mg['group']
mg['race_group'] = pd.Categorical(
    mg['race_group'], 
    categories=['NHW-HRL','NHB-HRL',
'NHW-DCIS', 'NHB-DCIS', 
'NHW-BC','NHB-BC'
                # 'NHB | HRL',
                # 'NHB | DCIS', 
                # 'NHB | BC'
               ]
)
mg['race_group'].unique()
````

````python
celltypes = mg['celltype_level3'].unique().tolist()
celltypes
````

````python
from statannotations.Annotator import Annotator


y = 'NHB-BC Up All'
for x in celltypes:

    tmp = mg[
        
        (mg['celltype_level3'] == x)
    ]
    
    plt.figure(figsize=(1.5, 1))
    
    ax = sns.boxplot(
        data=tmp,
        x='race_group', 
        hue='race_group',            
        y=y,
        palette=palette_meta  
    )
    
    ax.set_ylabel('')
    ax.set_xlabel('')
    ax.set_title(x)
    
    ax.set_xticklabels('')
    
    # --- Define pairs to compare ---
    pairs = [
        ('NHW-HRL',  'NHB-HRL'),
        ('NHW-DCIS', 'NHB-DCIS'),
        ('NHW-BC',   'NHB-BC')
    ]
    
    # --- Initialize annotator ---
    annot = Annotator(
        ax,
        pairs,
        data=tmp,        
        x='race_group',
        y=y
    )
    
    # --- Apply the test and display results ---
    annot.configure(
        test='Mann-Whitney',
        text_format='star',
        loc='inside',
        line_height=0.02,
        verbose=1
    )
    
    annot.apply_and_annotate()
    
    plt.tight_layout()
    plt.show()
````
