# PC1-5, with all covars

```python
dfp
```

```python
import pandas as pd
import numpy as np
from scipy.stats import spearmanr, mannwhitneyu, f_oneway
from statsmodels.stats.multitest import multipletests
import matplotlib.pyplot as plt
import seaborn as sns

pcs = [f'PC{x}' for x in range(1, 6)]
covars = ['race-stage', 'age', 'diabetes', 'obesity']

pval_table = pd.DataFrame(index=pcs, columns=covars)

for cov in covars:
    for pc in pcs:
        y = dfp[pc]
        x = dfp[cov]

        if x.nunique() >= 3 and x.dtype == 'O':  # categorical >=3 cats
            groups = [y[x == cat] for cat in x.unique() if pd.notnull(cat)]
            stat, pval = f_oneway(*groups)
        elif x.nunique() == 2:  # binary
            vals = x.unique()
            g1 = y[x == vals[0]]
            g2 = y[x == vals[1]]
            stat, pval = mannwhitneyu(g1, g2, alternative='two-sided')
        else:  # continuous
            stat, pval = spearmanr(x, y, nan_policy='omit')
        pval_table.loc[pc, cov] = pval

# Convert to float
pval_table = pval_table.astype(float)
```

```python
fdr_table = pval_table.copy()
for cov in covars:
    fdrs = multipletests(pval_table[cov], method='fdr_bh')[1]
    fdr_table[cov] = fdrs

# Save tables
pval_table.to_csv('data/Figure1/PC_covariate_pval.csv')
fdr_table.to_csv('data/Figure1/PC_covariate_FDR.csv')
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load your table
pval_table = pd.read_csv('data/Figure1/PC_covariate_pval.csv', index_col=0)

# Melt the DataFrame for plotting
plot_df_melt = pval_table.reset_index().melt(
    id_vars=pval_table.index.name or 'index', 
    var_name='Covariate', value_name='pval'
)
plot_df_melt['neglog10_p'] = -np.log10(plot_df_melt['pval'])
plot_df_melt['color'] = np.where(
    plot_df_melt['pval'] < 0.05,
    plot_df_melt['neglog10_p'],
    np.nan
)

# Swap axes: PC on x, covariate on y
fig, ax = plt.subplots(figsize=(2.2, 1.5))

# Covariates in reverse order
covariates = list(pval_table.columns)[::-1]

# Plot dots
for _, row in plot_df_melt.iterrows():
    color = row['neglog10_p'] if row['pval'] < 0.05 else 'lightgray'
    scatter = ax.scatter(
        row['index'], row['Covariate'],
        s=row['neglog10_p'] * 80,
        c=[color] if isinstance(color, float) else color,
        edgecolor='k', marker='o'
    )

# Set axis labels and ticks
ax.set_yticks(covariates)
ax.set_yticklabels(['Race-Stage','Age','Diabetes','Obesity'][::-1], fontsize=6)
ax.invert_yaxis()
ax.set_xticks(pval_table.index)
ax.set_xticklabels(pval_table.index, fontsize=6)

ax.set_xlabel('')
ax.set_ylabel('')
ax.margins(x=0.2, y=0.2)

# Colorbar (upper right, smaller)
cax = fig.add_axes([.82, 0.7, 0.02, 0.2])  # adjust numbers as needed
cbar = plt.colorbar(scatter, cax=cax)
cbar.ax.tick_params(labelsize=6)
cbar.set_label('-log10(p-value)', fontsize=6)

# cbar = plt.colorbar(scatter, ax=ax, label='-log10(p-value)', fraction=0.04, pad=0.04)
# cbar.ax.tick_params(labelsize=6)
# cbar.set_label('-log10(p-value)', fontsize=6)

# --- Size legend (lower right, smaller) ---
size_values = [1, 2, 3]  # example neglog10_p values
for size in size_values:
    ax.scatter([], [], s=size * 80, c='lightgray', edgecolor='k', label=f"{size}")

legend = ax.legend(
    title='-log10(p)',
    loc='lower left',                # anchor from lower left corner of legend
    bbox_to_anchor=(1, 0.05),     # move right (x>1) and adjust vertical position
    fontsize=5,
    title_fontsize=5,
    frameon=False,
    markerscale=0.5,
    handletextpad=.8    
)

plt.title('', fontsize=8)
plt.tight_layout()
plt.savefig('plots/Figure1/pval_PC1to5_covars.svg')
plt.show()
```

```python
# from statannotations.Annotator import Annotator
from scipy.stats import mannwhitneyu


var_p = 'diabetes'
color_set = {
    'N':'Gray',
    'Y':'#F472B6'
# 'Diabetes':'#F472B6',
# 'Obesity':'#F59E42'

}


for yname in ['PC3'
    ]:
    plt.figure(figsize=(1,1.5))
    ax = sns.boxplot(
        data=dfp,
        x=var_p,
        y=yname,
        hue=var_p,
        palette=color_set,
        flierprops=dict(marker='o', markersize=2, alpha=0.7)
    )
    sns.stripplot(data=dfp, x=var_p, y=yname, size=2, color='black', alpha=.2, ax=ax)

    # Assume var_p gives the group (should be exactly 2 values: e.g., 'N' and 'Y')
    group_labels = [g for g in dfp[var_p].unique() if pd.notnull(g)]
    if len(group_labels) == 2:
        group1, group2 = group_labels
        vals1 = dfp.loc[dfp[var_p] == group1, yname]
        vals2 = dfp.loc[dfp[var_p] == group2, yname]
        stat, p_value = mannwhitneyu(vals1, vals2, alternative='two-sided')
        ax.text(
            0.5, 1.01, f'p = {p_value:.2g}',
            ha='center', va='bottom',
            transform=ax.transAxes, fontsize=7
        )
    else:
        ax.text(
            0.5, 1.01, "Error: not two groups",
            ha='center', va='bottom',
            transform=ax.transAxes, fontsize=7
        )

    # --- Mann-Whitney pairwise comparisons ---
    # pairs = [
    #     ('NHW-HRL', 'NHB-HRL'), ('NHW-DCIS', 'NHB-DCIS'), ('NHW-BC', 'NHB-BC')
    # ]
    # annotator = Annotator(ax, pairs, data=dfp, x=var_p, y=yname)
    # annotator.configure(
    #     test='Mann-Whitney',
    #     text_format='star',
    #     loc='inside',
    #     comparisons_correction=None
    # )
    # annotator.apply_and_annotate()

    plt.xticks([0, 1], ['No', 'Diabetes'], fontsize=6)
    plt.xlabel('')
    ax.set_ylabel("PC3", labelpad=-3)
    plt.tight_layout()
    plt.savefig(f'plots/Figure1/PC_groupbyDiabetes_{yname}.svg')
```
