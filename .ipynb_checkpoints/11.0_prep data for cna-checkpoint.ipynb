```r
def read_gmt(file_path):
    gene_sets = {}
    with open(file_path, 'r') as f:
        for line in f:
            entries = line.strip().split('\t')
            gene_set_name = entries[0]
            description = entries[1]
            genes = entries[2:]
            gene_sets[gene_set_name] = {
                'description': description,
                'genes': genes
            }
    return gene_sets
```

# Export UMAP within each major cell type

```r
cellname = 'B'

raw <- qread(glue("data/03_cleandata/ft_{cellname}.qs"))
raw
```

```r
umap_coords <- Embeddings(raw, "umap")

# Optionally turn into a data frame and add cell IDs
umap_df <- as.data.frame(umap_coords)
umap_df$cell_id <- rownames(umap_df)

dim(umap_df)
umap_df[1:5,]
```

```r
plot(
  umap_df$umap_1,
  umap_df$umap_2,
  xlab = "UMAP 1",
  ylab = "UMAP 2",
  pch = 16,
  col = "steelblue",
  main = "UMAP Scatter Plot"
)
```

```r
for (cellname in c('B', 'CD4 T','CD8 T','other T', 'DC','Mono','NK')) {

    raw <- qread(glue("data/03_cleandata/ft_{cellname}.qs"))
    umap_coords <- Embeddings(raw, "umap")
    umap_df <- as.data.frame(umap_coords)
    
    write.csv(umap_df, 
              file = glue("data/03_cleandata/ft_{cellname}_UmapCoords.csv"), 
              row.names = TRUE)
}
```
