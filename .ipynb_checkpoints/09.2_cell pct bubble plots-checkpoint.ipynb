```python
meta = pd.read_csv('data/01_meta/meta_pt.csv', index_col=0)
#meta.index = [x.replace('_','-') for x in meta.index]

meta['race-stage'] = meta['race'].astype(str) + '-' + meta['group'].astype(str)
meta['race-stage'] = pd.Categorical(
    meta['race-stage'],
    categories=['NHW-HRL', 'NHW-DCIS', 'NHW-BC','NHB-HRL', 'NHB-DCIS', 'NHB-BC'],
    ordered=True
    )

meta
```

# Cell proportion

```python
dat = sc.read_h5ad(
    'data/03_cleandata/ft_all.h5ad'
)

dat
```

```python
dat.obs = pd.merge(dat.obs, meta[['dataset_name','sampleid']], 
                   on='dataset_name', how='left')
```

```python
meta_ordered = meta.sort_values([
    'race-stage','age'
]).reset_index(drop=True)
    
pt_ordered = meta_ordered['sampleid'].tolist()
meta_ordered
```

```python
ct = pd.crosstab(dat.obs['celltype_level3'], 
                   dat.obs['sampleid'])
ct = ct.loc[:, pt_ordered]

pct = ct.div(ct.sum(axis=0)) * 100 #.sum(axis=0)
pct
```

```python
pct.values.flatten().max()
```

```python
df = pct.reset_index().melt(id_vars='celltype_level3', var_name='Sample', value_name='Percent')

plt.figure(figsize=(6, 2.8))
ax = sns.scatterplot(
    data=df,
    x='Sample', y='celltype_level3',
    size='Percent', sizes=(3, 40),  # Adjust dot size range as needed
    hue='Percent', palette='viridis', legend='brief',
    size_order=[0, 30, 50] 
)

# ax.legend(bbox_to_anchor=(1.02, 0.5), loc='center left', borderaxespad=0, title='Percent')
plt.tight_layout(rect=[0, 0, 0.97, 1])  # Leave more room on right for legend
ax.margins(x=0.02, y=0.02)


ax.set_xlabel('')
ax.set_ylabel('')
plt.xticks(rotation=90)
plt.title('')
plt.tight_layout()

def add_bracket_below_xticks(ax, start_idx, end_idx, height=0.03, yoffset=-0.1, text=None, fontsize=8):
    """
    Add a bracket below the x-axis from start_idx to end_idx (inclusive).
    """
    # Convert x indices to data coords
    x0, x1 = start_idx, end_idx
    # y in axes fraction (0 is bottom)
    y = yoffset
    # Get axis transform for mixed coordinates: x=data, y=axes fraction
    trans = ax.get_xaxis_transform()
    # Draw bracket
    ax.plot([x0, x0, x1, x1], [y, y-height, y-height, y], color='k', lw=1, clip_on=False, transform=trans)
    if text:
        ax.text((x0+x1)/2, y-1.5*height, text, ha='center', va='top', fontsize=fontsize, transform=trans)


add_bracket_below_xticks(ax, 0, 9, height=0.02, yoffset=-0.15, text='NHW-HRL')
add_bracket_below_xticks(ax, 10, 15, height=0.02, yoffset=-0.15, text='NHW-DCIS')
add_bracket_below_xticks(ax, 16, 28, height=0.02, yoffset=-0.15, text='NHW-BC')
add_bracket_below_xticks(ax, 29, 38, height=0.02, yoffset=-0.15, text='NHB-HRL')
add_bracket_below_xticks(ax, 39, 42, height=0.02, yoffset=-0.15, text='NHB-DCIS')
add_bracket_below_xticks(ax, 43, 55, height=0.02, yoffset=-0.15, text='NHB-BC')


def add_bracket_left_yticks(ax, start_idx, end_idx, xoffset=-0.1, width=0.03, text=None, fontsize=8):
    """
    Add a vertical bracket to the left of the y-axis from start_idx to end_idx.
    - start_idx, end_idx: integer indices of y-ticks (top=0).
    - xoffset: how far left (axes fraction, negative means outside plot).
    - width: width of the bracket (axes fraction).
    """
    # y indices (in data coordinates)
    y0, y1 = start_idx, end_idx
    # x in axes fraction (0 is left edge)
    x = xoffset
    # Get axis transform for mixed coordinates: y=data, x=axes fraction
    trans = ax.get_yaxis_transform()
    # Draw vertical bracket
    ax.plot([x, x-width, x-width, x], [y0, y0, y1, y1], color='k', lw=1, clip_on=False, transform=trans)
    if text:
        ax.text(x-1.5*width, (y0+y1)/2, text, ha='right', va='center', fontsize=fontsize, transform=trans)


add_bracket_left_yticks(ax, 0, 3, xoffset=-0.18, width=0.01, text="B cells", fontsize=8) 
add_bracket_left_yticks(ax, 4, 8, xoffset=-0.18, width=0.01, text="CD4 T", fontsize=8) 
add_bracket_left_yticks(ax, 9, 11, xoffset=-0.18, width=0.01, text="CD8 T", fontsize=8) 
add_bracket_left_yticks(ax, 12, 14, xoffset=-0.18, width=0.01, text="Other T", fontsize=8)    
add_bracket_left_yticks(ax, 15, 17, xoffset=-0.18, width=0.01, text="DC", fontsize=8) 
add_bracket_left_yticks(ax, 18, 21, xoffset=-0.18, width=0.01, text="Monocyte", fontsize=8)   
add_bracket_left_yticks(ax, 22, 24, xoffset=-0.18, width=0.01, text="NK", fontsize=8) 


leg = ax.legend(
    loc='lower left',
    bbox_to_anchor=(-.2, -.2),  # push to bottom-left outside plot
    frameon=False,
    ncol=5, columnspacing=1, handletextpad=0.2,
    title='Percent'
)

plt.savefig('plots/FigureS2/cellpct.svg', bbox_inches='tight')
```
