```python
from matplotlib_venn import venn2
import matplotlib.pyplot as plt

def draw_venn_and_print(ax, A, B, left_name="Left", right_name="Right", title=""):
    # Draw venn from sets so counts are auto-computed
    v = venn2(subsets=(A, B), set_labels=(left_name, right_name), ax=ax)

    # Optional: tweak label font sizes (counts + set labels)
    for _id in ['10','01','11']:
        lbl = v.get_label_by_id(_id)
        if lbl is not None:
            lbl.set_fontsize(7)
    for lbl in v.set_labels:
        if lbl is not None:
            lbl.set_fontsize(7)

    ax.set_title(title, fontsize=7, pad=-6)

    # Compute sections
    left_only  = sorted(A - B)
    right_only = sorted(B - A)
    overlap    = sorted(A & B)

    # Print values to console (or a log)
    print(f"\n=== {title} ===")
    print(f"[{left_name} ONLY] (n={len(left_only)}):")
    for x in left_only:
        print("  -", x)
    print(f"[{right_name} ONLY] (n={len(right_only)}):")
    for x in right_only:
        print("  -", x)
    print(f"[OVERLAP {left_name} ∩ {right_name}] (n={len(overlap)}):")
    for x in overlap:
        print("  -", x)

    return v

# --- Your data (as you had) ---
s1 = set(dep_sel[dep_sel['subgroup']=='NHW-DCIS vs HRL']['pathway'])
s2 = set(dep_sel[dep_sel['subgroup']=='NHW-BC vs DCIS']['pathway'])
s3 = set(dep_sel[dep_sel['subgroup']=='NHW-BC vs HRL']['pathway'])

s4 = set(dep_sel[dep_sel['subgroup']=='NHB-DCIS vs HRL']['pathway'])
s5 = set(dep_sel[dep_sel['subgroup']=='NHB-BC vs DCIS']['pathway'])
s6 = set(dep_sel[dep_sel['subgroup']=='NHB-BC vs HRL']['pathway'])

fig, axs = plt.subplots(figsize=(3.8, 1.3), ncols=3)
plt.subplots_adjust(wspace=0.25)

draw_venn_and_print(
    ax=axs[0],
    A=s1, B=s4,
    left_name="NHW", right_name="NHB",
    title="DCIS vs HRL"
)

draw_venn_and_print(
    ax=axs[1],
    A=s2, B=s5,
    left_name="NHW", right_name="NHB",
    title="BC vs DCIS"
)

draw_venn_and_print(
    ax=axs[2],
    A=s3, B=s6,
    left_name="NHW", right_name="NHB",
    title="BC vs HRL"
)

plt.savefig('plots/Figure3/Venn_UpPathways.svg', bbox_inches='tight')
plt.show()
```

## Overlap heatmap

```python
dep_sel = dep[   
    (dep['padj']<.05) &   
    (dep['NES']>0)
]


s11 = set(dep_sel[dep_sel['subgroup']=='NHW-DCIS vs HRL']['pathway'])
s12 = set(dep_sel[dep_sel['subgroup']=='NHW-BC vs DCIS']['pathway'])
s13 = set(dep_sel[dep_sel['subgroup']=='NHW-BC vs HRL']['pathway'])

s14 = set(dep_sel[dep_sel['subgroup']=='NHB-DCIS vs HRL']['pathway'])
s15 = set(dep_sel[dep_sel['subgroup']=='NHB-BC vs DCIS']['pathway'])
s16 = set(dep_sel[dep_sel['subgroup']=='NHB-BC vs HRL']['pathway'])


dep_sel = dep[   
    (dep['padj']<.05) &   
    (dep['NES']<0)
]


s21 = set(dep_sel[dep_sel['subgroup']=='NHW-DCIS vs HRL']['pathway'])
s22 = set(dep_sel[dep_sel['subgroup']=='NHW-BC vs DCIS']['pathway'])
s23 = set(dep_sel[dep_sel['subgroup']=='NHW-BC vs HRL']['pathway'])

s24 = set(dep_sel[dep_sel['subgroup']=='NHB-DCIS vs HRL']['pathway'])
s25 = set(dep_sel[dep_sel['subgroup']=='NHB-BC vs DCIS']['pathway'])
s26 = set(dep_sel[dep_sel['subgroup']=='NHB-BC vs HRL']['pathway'])
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt


group_palette = {
    # NHW comparisons - cold colors
    'NHW-DCIS vs HRL': '#91C4F2',  # light sky blue
    'NHW-BC vs HRL':   '#1F4E79',  # deep navy blue
    'NHW-BC vs DCIS':  '#00A6A6',  # vivid cyan-teal
    # NHB comparisons - warm colors
    'NHB-DCIS vs HRL': '#F4978E',  # salmon pink
    'NHB-BC vs HRL':   '#A32020',  # deep crimson red
    'NHB-BC vs DCIS':  '#F27F0C'   # vivid orange
}

def base_comp(name: str) -> str:
    """Strip '-up'/'-down' suffix to map into group_palette."""
    name = name.strip()
    if name.endswith('-up'):
        return name[:-3].strip()
    if name.endswith('-down'):
        return name[:-5].strip()
    return name

    
# ---- 1) Build wide overlap table (counts) ----
sets_dict = {
    'NHW-DCIS vs HRL-up':  s11,
    'NHW-BC vs DCIS-up': s12,
    'NHW-BC vs HRL-up':   s13,
    'NHB-DCIS vs HRL-up':  s14,
    'NHB-BC vs DCIS-up': s15,
    'NHB-BC vs HRL-up':   s16,

    'NHW-DCIS vs HRL-down':  s21,
    'NHW-BC vs DCIS-down': s22,
    'NHW-BC vs HRL-down':   s23,
    'NHB-DCIS vs HRL-down':  s24,
    'NHB-BC vs DCIS-down': s25,
    'NHB-BC vs HRL-down':   s26
}
keys = list(sets_dict.keys())

# overlap count matrix
overlap_mat = pd.DataFrame(0, index=keys, columns=keys, dtype=int)
for a in keys:
    for b in keys:
        overlap_mat.loc[a, b] = len(sets_dict[a] & sets_dict[b])

# ---- 2) Plot heatmap with annotations (counts) ----
plt.figure(figsize=(4, 2))
ax = sns.heatmap(
    overlap_mat,
    cmap="YlGnBu",
    annot=True, fmt="d",            # show integer counts
    linewidths=0.5, linecolor="white",
    cbar_kws={"shrink": 0.7}
)

# tidy labels
ax.set_xlabel("")
ax.set_ylabel("")
ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha='right', fontsize=5)
ax.set_yticklabels(ax.get_yticklabels(), rotation=0, fontsize=5)

# ---- Color tick labels by comparison group ----
# X ticks
for lbl in ax.get_xticklabels():
    key_txt = lbl.get_text()
    color = group_palette.get(base_comp(key_txt), 'black')
    lbl.set_color(color)
# Y ticks
for lbl in ax.get_yticklabels():
    key_txt = lbl.get_text()
    color = group_palette.get(base_comp(key_txt), 'black')
    lbl.set_color(color)

plt.tight_layout()
plt.savefig('plots/Figure3/overlapheatmap.svg', bbox_inches='tight')
plt.show()
```

## Overlap heatmap

```python
dep_sel = dep[   
    (dep['padj']<.05) &   
    (dep['NES']>0)
]


s1 = set(dep_sel[dep_sel['subgroup']=='HRL']['pathway'])
s2 = set(dep_sel[dep_sel['subgroup']=='DCIS']['pathway'])
s3 = set(dep_sel[dep_sel['subgroup']=='BC']['pathway'])


dep_sel = dep[   
    (dep['padj']<.05) &   
    (dep['NES']<0)
]


s4 = set(dep_sel[dep_sel['subgroup']=='HRL']['pathway'])
s5 = set(dep_sel[dep_sel['subgroup']=='DCIS']['pathway'])
s6 = set(dep_sel[dep_sel['subgroup']=='BC']['pathway'])
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt


group_palette = {
    'HRL-up': '#F5C16C',   # light gold
    'DCIS-up': '#EA7D30',  # orange
    'BC-up': '#7A491A',    # deep brown (darkest)
    'HRL-down': '#A3CCE9',   # lighter sky blue (soft and clear)
    'DCIS-down': '#2C8F91',  # teal
    'BC-down':   '#7B5BA5'   # muted indigo
}



# ---- 1) Build wide overlap table (counts) ----
sets_dict = {
    'HRL-up':  s1,
    'DCIS-up': s2,
    'BC-up':   s3,
    'HRL-down': s4,
    'DCIS-down': s5,
    'BC-down':  s6
}
keys = list(sets_dict.keys())

# overlap count matrix
overlap_mat = pd.DataFrame(0, index=keys, columns=keys, dtype=int)
for a in keys:
    for b in keys:
        overlap_mat.loc[a, b] = len(sets_dict[a] & sets_dict[b])


# ---- 2) Plot heatmap with annotations (counts) ----
plt.figure(figsize=(3, 1.5))
ax = sns.heatmap(
    overlap_mat,
    cmap="YlGnBu",
    annot=True, fmt="d",            # show integer counts
    linewidths=0.5, linecolor="white",
    cbar_kws={"shrink": 0.7}
)

# ---- Color tick labels by comparison group ----
# X ticks
for lbl in ax.get_xticklabels():
    key_txt = lbl.get_text()
    color = group_palette[key_txt]
    lbl.set_color(color)
# Y ticks
for lbl in ax.get_yticklabels():
    key_txt = lbl.get_text()
    color = group_palette[key_txt]
    lbl.set_color(color)

plt.tight_layout()
plt.savefig('plots/Figure4/overlapheatmap.svg', bbox_inches='tight')
plt.show()
```
