```python
from utils import *
```

```python
os.chdir('../')
```

```python
dft = pd.DataFrame()

idx = 0

for cell in ['Mono','DC','CD4T','B']:

    for dz in ['HRL','DCIS','BC']:

        idx += 1

        tmp = pd.read_csv(
            f'data/Tables/{cell}_{dz}.csv', 
            index_col=0
                     )

        tmp['Cell lineage'] = [cell] * tmp.shape[0]
    
        dft = tmp if idx==1 else pd.concat((dft, tmp))

dft
```

```python
dft = dft[['Cell lineage','analysis_group','Term','NES','NOM p-val', 'FDR q-val','Lead_genes']].reset_index(drop=True)
dft
```

```python
dft['Cell lineage'] = [
    'Monocyte' if x=='Mono' else
    'B cell' if x=='B' else
    'CD4 T cell' if x=='CD4T' else
    x for x in dft['Cell lineage']
]
```

```python
dft.columns = ['Cell lineage','Group','Term','NES','NOM p-val','FDR q-val','Lead_genes']
```

```python
dft.to_csv('data/Tables/ST_CNAgsea.csv')
```
