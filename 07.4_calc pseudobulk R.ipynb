```r
dat = qs::qread('data/03_cleandata/ft_all.qs')
dat
```

# Calc pseudobulk

## Donor

```r
pb <- AggregateExpression(dat, 
                           return.seurat = T, 
                           group.by = c("dataset_name"))

pb
```

```r
qs::qsave(pb, 
          'data/06_pb/pb_donor.qs')
```

```r
norm_mat <- GetAssayData(pb, assay = "RNA", slot = "data")  # log-normalized
norm_df <- as.data.frame(as.matrix(norm_mat))
write.csv(norm_df, file = "data/06_pb/pb_donor_normalizedexpression.csv", quote = FALSE)
```
