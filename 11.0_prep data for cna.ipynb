# Export UMAP within each major cell type

```r
for (cellname in c('B', 'CD4 T','CD8 T','other T', 'DC','Mono','NK')) {

    raw <- qread(glue("data/03_cleandata/ft_{cellname}.qs"))
    umap_coords <- Embeddings(raw, "umap")
    umap_df <- as.data.frame(umap_coords)
    
    write.csv(umap_df, 
              file = glue("data/03_cleandata/ft_{cellname}_UmapCoords.csv"), 
              row.names = TRUE)
}
```
